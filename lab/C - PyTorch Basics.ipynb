#### 1. Converting NumPy arrays to PyTorch tensors

A torch.Tensor is a multi-dimensional matrix containing elements of a single data type. Calculations between tensors can only happen if the tensors share the same dtype. In some cases tensors are used as a replacement for NumPy to use the power of GPUs (more on this later). Let's try to initialize a numpy array first!


````python
import torch
import numpy as np
import sys
````

````python
arr = np.array([1,2,3,4,5]).astype('float64')
print(arr)
print(arr.dtype)
print(type(arr))
````

````python
x = torch.from_numpy(arr)
# Equivalent to x = torch.as_tensor(arr)
print(x)
````

````python
# Print the datatype
print(x.dtype)
````

````python
# Print the tensor object type
print(type(x))
print(x.type()) # this is more specific!
````

````python
arr2 = np.arange(0.,12.).reshape(4,3)
print(arr2)
print(type(arr2))
````

````python
x2 = torch.from_numpy(arr2)
print(x2)
print(x2.type())
````

````python
rows, cols = 5, 5

# Create matrix using array comprehension
mat = np.matrix([[np.round(np.random.random(), 3) for _ in range(cols)] for _ in range(rows)])
print(mat)
print(type(mat))


xmat = torch.from_numpy(mat)
print(xmat)
print(type(xmat))
````

````python
# Check if an object is a tensor
torch.is_tensor(mat)
````

#### 5. Tensor Arithmetic

Adding tensors can be performed a few different ways depending on the desired result.

As a simple expression:


````python
a = torch.tensor([1,2,3], dtype=torch.float)
b = torch.tensor([4,5,6], dtype=torch.float)
print(a + b)
````

As arguments passed into a torch operation:


````python
print(torch.add(a, b))
````

With an output tensor passed in as an argument:


````python
result = torch.empty(3)
torch.add(a, b, out=result)  # equivalent to result=torch.add(a,b)
print(result)
````

````python
a.add_(b)  # equivalent to a=torch.add(a,b)
print(a)
````

**NOTE:** Any operation that changes a tensor in-place is post-fixed with an underscore _.

In the above example: `a.add_(b)` changed `a`.


#### Basic Tensor Operations

**Arithmetic**

| Operation | Function | Description |
|---|---|---|
| a + b | `a.add(b)` | element wise addition |
| a - b | `a.sub(b)` | subtraction |
| a * b | `a.mul(b)` | multiplication |
| a / b | `a.div(b)` | division |
| a % b | `a.fmod(b)` | modulo (remainder after division) |
| $a^{b}$ | `a.pow(b)` | power |

**Monomial operations**

| Operation | Function | Description |
|---|---|---|
| $\lvert a \rvert$ | `torch.abs(a)` | absolute value |
| 1/a | `torch.reciprocal(a)` | reciprocal |
| $\sqrt{a}$ | `torch.sqrt(a)` | square root |
| log(a) | `torch.log(a)` | natural log |
| $e^{a}$ | `torch.exp(a)` | exponential |
| 12.34  ==>  12. | `torch.trunc(a)` | truncated integer |
| 12.34  ==>  0.34 | `torch.frac(a)` | fractional component |

**Summary statistics**

| Operation | Function | Description |
|---|---|---|
| $\sum a$ | `torch.sum(a)` | sum |

`torch.max(a, b)` returns a tensor the size of `a` containing the element-wise max of `a` and `b`.

**NOTE:**  Most arithmetic operations require float values. Those that do work with integers return integer tensors.

For example, `torch.div(a,b)` performs floor division (truncates the decimal) for integer types, and classic division for floats.


#### 6. Dot Products

A [dot product](https://en.wikipedia.org/wiki/Dot_product) is the sum of the products of the corresponding entries of two 1D tensors. If the tensors are both vectors, the dot product is given as:

$\begin{bmatrix} a & b & c \end{bmatrix} \;\cdot\; \begin{bmatrix} d & e & f \end{bmatrix} = ad + be + cf$

If the tensors include a column vector, then the dot product is the sum of the result of the multiplied matrices. For example:

$\begin{bmatrix} a & b & c \end{bmatrix} \;\cdot\; \begin{bmatrix} d \\ e \\ f \end{bmatrix} = ad + be + cf$

Dot products can be expressed as [`torch.dot(a,b)`](https://pytorch.org/docs/stable/torch.html#torch.dot) or `a.dot(b)` or `b.dot(a)`


````python
a = torch.tensor([1,2,3], dtype=torch.float)
b = torch.tensor([4,5,6], dtype=torch.float)
print(a.dot(b))
````

**NOTE:**  There's a slight difference between `torch.dot()` and `numpy.dot()`. While `torch.dot()` only accepts 1D arguments and returns a dot product, `numpy.dot()` also accepts 2D arguments and performs matrix multiplication. We show matrix multiplication below.


#### 7. Matrix Multiplication

2D [Matrix multiplication](https://en.wikipedia.org/wiki/Matrix_multiplication) is possible when the number of columns in tensor `A` matches the number of rows in tensor `B`. In this case, the product of tensor `A` with size $(x,y)$ and tensor `B` with size $(y,z)$ results in a tensor of size $(x,z)$

$\begin{bmatrix} a & b & c \\
d & e & f \end{bmatrix} \;\times\; \begin{bmatrix} m & n \\ p & q \\ r & s \end{bmatrix} = \begin{bmatrix} (am+bp+cr) & (an+bq+cs) \\
(dm+ep+fr) & (dn+eq+fs) \end{bmatrix}$

Matrix multiplication can be computed using [`torch.mm(a,b)`](https://pytorch.org/docs/stable/torch.html#torch.mm) or `a.mm(b)` or `a @ b`


````python
a = torch.tensor([[0,2,4],[1,3,5]], dtype=torch.float)
b = torch.tensor([[6,7],[8,9],[10,11]], dtype=torch.float)

print('a: ',a.size())
print('b: ',b.size())
print('a x b: ',torch.mm(a,b).size())
````

````python
print(torch.mm(a,b))
````

````python
print(a.mm(b))
````

````python
print(a @ b)
````

### Laboratory Task 5 — Solution

````python
# 1. Perform standard imports
import torch
import numpy as np
````

````python
# 2. set_seed(): seeds NumPy and PyTorch, returns nothing
def set_seed(seed: int) -> None:
    np.random.seed(seed)
    torch.manual_seed(seed)
````

````python
# 3. Six random integers in [0, 5) with seed 42
set_seed(42)
arr = np.random.randint(0, 5, size=6).astype('int32')

assert np.array_equal(arr, np.array([3, 4, 2, 4, 4, 1])), "Oops"
print(arr)
````

````python
# 4. Tensor x from the array (shares memory with arr)
x = torch.from_numpy(arr)
print(x, x.dtype)
````

````python
# 5. Change the dtype of x from int32 to int64
print('Old:', x.type())
x = x.type(torch.int64)
print('New:', x.type())
````

````python
# 6. Reshape x into a 3x2 tensor (x.view(3, 2) or x.reshape(3, -1) also work)
x = x.reshape(3, 2)
print(x)
````

````python
# 7. Right-hand column of x (slicing with 1: keeps it as a 3x1 column)
print(x[:, 1:])
````

````python
# 8. Square values of x without changing x.
#    x.pow(2) returns a NEW tensor; the in-place version would be x.pow_(2).
x_squared = x.pow(2)          # equivalently x ** 2 or torch.mul(x, x)
print(x_squared)
print('x is unchanged:\n', x)
````

````python
# 9. y has 6 elements like x, and shape (2, 3) so that (3x2) @ (2x3) is defined
set_seed(42)
y = torch.randint(0, 5, (2, 3))

assert torch.equal(y, torch.tensor([[2, 2, 1], [4, 1, 0]])), "Oops"
print(y)
````

````python
# 10. Matrix product of x and y: (3x2) @ (2x3) -> (3x3)
prod = torch.mm(x, y)
assert torch.equal(prod, x @ y)
prod
````

**Notes.**

- `torch.from_numpy` shares memory with `arr`. Changing the dtype in step 5 makes a copy, so
  later steps no longer touch `arr`.
- `x.pow(2)` is not in place (in-place methods end in `_`, like `pow_`). That is why `x` prints
  unchanged after step 8. The original template computed `x.pow(2)` but discarded the result.
- Matrix multiplication needs the inner dimensions to agree. With `x` of shape $3\times2$, `y` must
  be $2\times n$. Using $2\times3$ keeps the same six elements and gives a $3\times3$ product:
  for example, entry $(1,1)$ is $3\cdot2 + 4\cdot4 = 22$.
